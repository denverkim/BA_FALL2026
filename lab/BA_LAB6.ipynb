```python
x = [1, 5, 1.5, 8, 1, 9]
x
```

```python
y = [2, 8, 1.8, 8, 0.6, 11]
y
```

```python
import matplotlib.pyplot as plt
plt.scatter(x, y)
plt.grid()
plt.show()
```

```python
import numpy as np

X = np.array([x,y])
X = X.T
X
```

```python
from sklearn.cluster import KMeans

kmeans = KMeans(n_clusters=2)
kmeans.fit(X)
```

```python
center = kmeans.cluster_centers_
label = kmeans.labels_
```

```python
center
```

```python
# 중앙값 표시
plt.scatter(x, y, c=label)
plt.grid()
plt.scatter(center[:,0], center[:,1], s=300, c='red', marker='x')
plt.show()
```

```python
from scipy.cluster.hierarchy import dendrogram, linkage

dendrogram(linkage(X, 'ward'))
```
