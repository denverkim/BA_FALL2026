# 인코딩 투토리얼

```python
data = {
        'country':['france', 'spain', 'germany', 'spain', 'germany', 'france', 'spain', 'france', 'germany', 'france'],
              'age':[44, 27, 30, 38, 40, 35,66, 48, 50, 37],

        'salary':[72000, 48000, 54000, 61000, 55000, 58000, 52000, 79000, 83000, 67000],
        'purchased':['no', 'yes', 'no', 'no', 'yes', 'yes', 'no', 'yes', 'no', 'yes']
        }
df = pd.DataFrame(data)
df
```

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
df.country = le.fit_transform(df.country)
df
```

```python
le.classes_ #범주 확인할 때
```

```python
#get_dummy
dummy = pd.get_dummies(df.purchased)
dummy
```

```python
df = pd.concat([df, dummy], axis=1)
df
```

```python
df = df.drop(['no', 'purchased'], axis=1)
df
```

```python
df.rename(columns={'yes':'purchased'}, inplace=True)
df
```

```python
from sklearn.preprocessing import OneHotEncoder
ohe = OneHotEncoder()
dummy = ohe.fit_transform(df[['country', 'purchased']]).toarray()
dummy_df = pd.DataFrame(dummy, columns=['france', 'germany', 'spain', 'no', 'yes'])
dummy_df
```

```python
df = pd.concat([df, dummy_df], axis=1)
df
```

# 세일 데이터 전처리 실습

```python
# 훈련, 테스트 데이터를 임포트한 후에 기본 데이터 탐색 수행

import pandas as pd

train = pd.read_csv('/content/bigmart_train.csv')
train
```

```python
# 데이터 사이즈 확인
train.shape
```

```python
# 데이터 타입 확인
train.dtypes
```

```python
# 기초 통계량
train.describe()
```

```python
# 널값 검색
train.isna().sum()
```

```python
# 테스트 테이터에 아웃렛세일 열을 만들고 1을 할당
test = pd.read_csv('/content/bigmart_test.csv')
test['Item_Outlet_Sales'] = 1
test
```

```python
# Combine train and test data. 훈련과 테스트 데이터를 합침
df = pd.concat([train, test])
df.shape
```

```python
# 분실값 및 매칭되지 않는 값 대체
# 1. 아이템 무게의 분실값을 중위수로 대체
df['Item_Weight'].fillna(df['Item_Weight'].median(), inplace=True)
```

```python
# 널값 확인
df.isna().sum()
```

```python
# Impute missing values by median for item visibility. 노출도 0인 값을 중위수로 대체
df.loc[df['Item_Visibility']==0, 'Item_Visibility'] = df.Item_Visibility.median()
```

```python
df.describe()
```

```python
# 아웃렛 사이즈 빈도
df['Outlet_Size'].value_counts(dropna=False)
```

```python
# 아웃렛 사이즈의 분실값을 Other로 대체
df.Outlet_Size.fillna('Other', inplace=True)
```

```python
# 아웃렛 사이즈 빈도 체크
df.Outlet_Size.value_counts()
```

```python
# 지방 함유율 빈도
df['Item_Fat_Content'].value_counts()
```

```python
# 지방함유율의 값 대체
# LF to Low Fat, reg to Regular, low fat to Low Fat
df['Item_Fat_Content'] = df['Item_Fat_Content'].replace(['LF', 'low fat'], 'Low Fat')
df['Item_Fat_Content'] = df['Item_Fat_Content'].replace('reg', 'Regular')
```

```python
# 바꿨는지 확인
df.Item_Fat_Content.value_counts()
```

```python
# Year라는 새로운 열 만들기
df['Year'] = 2026 - df['Outlet_Establishment_Year']
df.head()
```

```python
# 필요없는 열을 삭제 (exclude item identifier, outlet identifier, item fat content, outlet establishment year, and item type).
df.drop(['Item_Identifier', 'Outlet_Identifier', 'Item_Fat_Content', 'Outlet_Establishment_Year', 'Item_Type'], axis=1, inplace=True)
df.head()
```

```python
# 범주형 변수를 인코딩
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
df['Outlet_Size'] = le.fit_transform(df['Outlet_Size'])
df['Outlet_Location_Type'] = le.fit_transform(df['Outlet_Location_Type'])
df['Outlet_Type'] = le.fit_transform(df['Outlet_Type'])
df.head()
```
